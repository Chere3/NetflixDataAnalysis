# Preprocesado de datos del contenido multimedia de netflix
Ahora, una vez identificados los problemas y caracteristias de este dataset, se prepararan los datos para el análisis estádistico o modelado de los mismos


```python
import pandas as pd

original_df = pd.read_csv("../../data/raw/netflix_data.csv")
df = original_df.copy()
```

### Procesamiento de valores nulos
Todos los datos faltantes serán llenados con datos ya sean como la media, average u otros

```python
df["cast"] = df["cast"].fillna("Desconocido")
df["country"] = df["country"].fillna("Desconocido")
df["date_added"] = df["date_added"].fillna("January 01, 1970")
df["description"] = df["description"].fillna("Sin descripción")
df["director"] = df["director"].fillna("Desconocido")
df["duration"] = df["duration"].fillna(1)
df["genre"] = df["genre"].fillna("Desconocido")
df["release_year"] = df["release_year"].fillna(1970)
df["show_id"] = df["show_id"].fillna("S0")
df["title"] = df["title"].fillna("Desconocido")
df["type"] = df["type"].fillna("Desconocido")
```

## Obtener outliners
Obtener todos los datos que se salgan del rango medio de datos, para evitar datos fuera del rango preciso de series y peliculas, con el tal de tener graficas más estandarizadas

### Obtener todos los datos fuera usando el rango intercualitico (IQR)
Todos los datos fuera del rango intercualitico tienen que ser eliminados con el tal de mantener las graficas en su punto

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Teniendo en cuenta que el formateo de los datos es asi
sns.scatterplot(df["release_year"])
plt.show()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], format='%B %d, %Y', errors='coerce')
df["genre"] = df["genre"].astype("category")
df["country"] = df["country"].astype("category")
df["type"] = df["type"].astype("category")
df["release_year"] = pd.to_numeric(df["release_year"])
```
